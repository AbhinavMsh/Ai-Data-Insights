```python
# --- Setup / Import ---
import numpy as np
```

# --- Section: Theory Examples ---

```python
# np.zeros: 2D array
np.zeros((2, 3))
```

```python
# np.arange: 1D range 0 to 9
np.arange(10)
```

```python
# np.arange: with float data type
np.arange(2, 10, dtype=float)
```

```python
# np.arange: with step size
np.arange(2, 3, 0.1)
```

```python
# np.linspace: evenly spaced numbers
np.linspace(1., 4., 6)
```

```python
# np.indices: grid indices
np.indices((3, 3))
```

# --- Section: Array Creation Examples ---

```python
# np.array: create array from python list
marks = np.array([75, 82, 90, 68, 88])
print(marks)
```

```python
# np.zeros: 1D array of zeros
a = np.zeros(5)
print(a)
```

```python
# np.zeros: 2D array of zeros
a = np.zeros((2, 3))
print(a)
```

```python
# np.ones: 2D array of ones
a = np.ones((2, 3))
print(a)
```

```python
# np.full: array filled with a specific value (7)
a = np.full((2, 3), 7)
print(a)
```

```python
# np.arange: range from 1 to 10
a = np.arange(1, 11)
print(a)
```

```python
# np.arange: range with step size of 5
a = np.arange(0, 20, 5)
print(a)
```

```python
# np.linspace: 5 evenly spaced values between 0 and 10
a = np.linspace(0, 10, 5)
print(a)
```

```python
# np.eye: 3x3 identity matrix
a = np.eye(3)
print(a)
```
